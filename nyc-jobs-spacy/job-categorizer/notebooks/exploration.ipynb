# Job Categorizer Exploration

This notebook is used for exploratory data analysis (EDA) on the job data. We will visualize the job descriptions, business titles, and civil service titles to understand their distributions and relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set the style of seaborn
sns.set(style='whitegrid')
```

```python
# Load the job data
data_path = '../data/raw/jobs.csv'
jobs_df = pd.read_csv(data_path)

# Display the first few rows of the dataframe
jobs_df.head()
```

```python
# Visualize the distribution of business titles
plt.figure(figsize=(12, 6))
sns.countplot(y='business_title', data=jobs_df, order=jobs_df['business_title'].value_counts().index)
plt.title('Distribution of Business Titles')
plt.xlabel('Count')
plt.ylabel('Business Title')
plt.show()
```

```python
# Visualize the distribution of civil service titles
plt.figure(figsize=(12, 6))
sns.countplot(y='civil_service_title', data=jobs_df, order=jobs_df['civil_service_title'].value_counts().index)
plt.title('Distribution of Civil Service Titles')
plt.xlabel('Count')
plt.ylabel('Civil Service Title')
plt.show()
```

```python
# Visualize the length of job descriptions
jobs_df['description_length'] = jobs_df['job_description'].apply(len)
plt.figure(figsize=(12, 6))
sns.histplot(jobs_df['description_length'], bins=30, kde=True)
plt.title('Distribution of Job Description Lengths')
plt.xlabel('Length of Job Description')
plt.ylabel('Frequency')
plt.show()
```
