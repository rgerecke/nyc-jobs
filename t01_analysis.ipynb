# First Analysis
Trying to use SpaCy's text categorization on the existing database.

## Installation

```python
import polars as pl
import spacy

df = pl.read_csv("database.csv")
nlp = spacy.load("en_core_web_sm")
```

At some point we should to an actual train/test/split, but for now I'm just going to sample 1000 documents to get a sense of how everything works.

This part also creates a column called `comms` with a flag for whether or not the `job_category` includes "Communications" (indicating that the job is related to Communications work). In the future, we will want to tag which jobs are AI-related or adjacent, which will be a custom labelling situation. For now, we work with the labelled data we already have.

```python
df_mut = (
    df.with_columns(
        pl.col('job_category')
            .str.contains("Communications")
            .alias('comms')
    )
        .sample(n=1000,seed=47)
)

df_mut.get_column('comms').value_counts()
```
